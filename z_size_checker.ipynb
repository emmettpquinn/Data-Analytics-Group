```python
import pandas as pd
import re

def clean_address(addr):
    if pd.isna(addr):
        return addr
    
    addr = addr.upper()                 # uppercase
    addr = addr.replace(",", " ")       # remove commas
    addr = re.sub(r"[^A-Z0-9\s]", " ", addr)  # remove weird symbols
    addr = re.sub(r"\s+", " ", addr)    # remove double spaces
    addr = addr.strip()
    
    return addr

df = pd.read_csv("ppr-group-22312913-train.csv")
df["address_clean"] = df["Address"].apply(clean_address)
```

```python
from collections import Counter

tokens = df["address_clean"].str.split().explode()

token_counts = Counter(tokens)

top_tokens = pd.DataFrame(token_counts.most_common(50),
                          columns=["token","count"])

print(top_tokens)
```

```python
apt_words = ["APT","APARTMENT","UNIT","FLAT"]
house_words = ["HOUSE","BUNGALOW","LODGE","VILLA"]

def property_type(addr):
    words = addr.split()
    
    if any(w in words for w in apt_words):
        return "apartment"
    
    if any(w in words for w in house_words):
        return "house"
    
    return "unknown"

df["property_type_guess"] = df["address_clean"].apply(property_type)
```
