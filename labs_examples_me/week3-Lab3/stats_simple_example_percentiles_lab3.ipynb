```python
# Import package numpy for numeric computing
import numpy as np

# Import package matplotlib for visualisation/plotting
import matplotlib.pyplot as plt

#For showing plots directly in the notebook run the command below
%matplotlib inline
```

```python

a = np.array([30000, 30000, 30000, 30000, 30000, 30000, 30000, 430000])
print("a: " + str(a))

min_a = np.min(a)
print("min: " + str(min_a))

p_25 = np.percentile(a, 25)
print("1st q:" + str(p_25))

median_a = np.median(a)
print("median: " + str(median_a))

p_50 = np.percentile(a, 50)
print("2nd q:" + str(p_50))

p_75 = np.percentile(a, 75)
print("3rd q:" + str(p_75))

max_a = np.max(a)
print("max: " + str(max_a))

mean_a = np.mean(a)
print("\nmean: " + str(mean_a))
```

```python
a = np.array([140, 140, 145, 149, 150, 151, 157, 163, 229])
print("a: " + str(a))

min_a = np.min(a)
print("min: " + str(min_a))

p_25 = np.percentile(a, 25)
print("1st q:" + str(p_25))

median_a = np.median(a)
print("median: " + str(median_a))

p_50 = np.percentile(a, 50)
print("2nd q:" + str(p_50))

p_75 = np.percentile(a, 75)
print("3rd q:" + str(p_75))

max_a = np.max(a)
print("max: " + str(max_a))

mean_a = np.mean(a)
print("\nmean: " + str(mean_a))
```

```python
p_25 = np.percentile(a, 25, interpolation='linear')
print("25perc linear:" + str(p_25))

p_25 = np.percentile(a, 25, interpolation='lower')
print("25perc lower:" + str(p_25))

p_25 = np.percentile(a, 25, interpolation='higher')
print("25perc higher:" + str(p_25))

p_25 = np.percentile(a, 25, interpolation='nearest')
print("25perc nearest:" + str(p_25))

p_25 = np.percentile(a, 25, interpolation='midpoint')
print("25perc midpoint:" + str(p_25))

```
