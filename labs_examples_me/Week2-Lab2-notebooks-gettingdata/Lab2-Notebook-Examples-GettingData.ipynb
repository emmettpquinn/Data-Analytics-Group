### Example 4: Parsing News Articles with Newspaper3k

The `newspaper3k` library simplifies extracting structured information from news articles (title, authors, publication date, text). This is much easier than manually parsing HTML with BeautifulSoup for common news websites.

````python
# we can download a newsarticle and parse it using the newspaper3k library
# https://buildmedia.readthedocs.org/media/pdf/newspaper/latest/newspaper.pdf
# newspaper cannot parse all types of html files, for more complex file structure we still need 'beautifulsoup'
from newspaper import Article

url ="https://www.irishtimes.com/politics/2025/01/23/new-government-pledges-to-go-on-offensive-over-us-trade-alongside-reforms-in-housing-justice-and-climate/"
article = Article(url)
article.download()

#print(article.html)

article.parse()
print("Authors:", article.authors)
print("Date:", article.publish_date)
print("Title:", article.title)
print("Text:", article.text)
print("\nURL:", article.url)
````

### Example 5: Working with RSS Feeds (XML Format)

RSS feeds are XML files that list recently published content from websites. We can parse these feeds to extract URLs of articles, then download and parse each article. This is useful for monitoring news sources or blogs programmatically.

**Important:** When scraping websites, always:
- Add delays between requests (`time.sleep()`) to avoid overloading servers
- Respect the website's `robots.txt` file
- Consider the website's Terms of Service

````python
#Downloading and working with an XML file
#Get the whole RSS feed for the Irish Times news articles
#This is an XML file listing the URLs of individual news articles published online
#Need to know the structure of the XML to be able to extract text from specific tags

#Parse the XML file to retrieve the URLs for individual news articles.
#Parse each article's HTML page

def getArticleDetailsByUrl(url):
    article = Article(url)
    article.download()

    #print(article.html)
    article.parse()
    authors = article.authors
    date = article.publish_date
    title = article.title
    text = article.text

    return [authors, date, title, text]

def scrapeRSSFeed(rss_feed):
    d = feedparser.parse(rss_feed)
    #print(d)
    #print(d['entries'], "\n")
        
    for item in d['entries']:
        #Extract an article URL
        article_url = item['link']
        print(article_url)
        try:
            [authors, date, title, text] = getArticleDetailsByUrl(article_url)
            print("\nArticle title:", title, "\n")
            print("\nArticle first paragraph:", text.split("\n")[0], "\n")
            #we introduce a delay after each article download to avoid overloading the IrishTimes server
            time.sleep(5)
        except requests.RequestException as e:
            print("[Error]: " + str(e))
            
#Here you have your very own RSS feed reader in a few lines of code.
if __name__ == '__main__':

    #The URL of the XML file
    url='https://www.irishtimes.com/rss/irish-times-top-10-stories-1.4019566'
    xml_page = requests.get(url).text
    
    #Look at the structure of the XML file
    #To have a proper look, open the XML file with a text editor
    print(xml_page[:1000])

    # Call the method that parses a given XML file
    scrapeRSSFeed(url)
````

````python
#Get a PDF file, save it to disk.

# Give url of the PDF file
url='http://www.greenteapress.com/thinkpython/thinkpython.pdf'
# Download the pdf file into request_object
request_object = requests.get(url)

#PDF is a binary format. Use request.content instead of request.text
#Write binary content on your machine's disk in a file named 'thinkpython.pdf'
with open("thinkpython.pdf", "wb") as pdffile:
    # Look at the conent of the file; it looks all gibberish since it is a binary format.
    # To make sense of the content, we need tools that can read pdf format and extract it to plain text.
    # See next cell for pdftotext tool.
    print(request_object.content[:500])
    
    #Print the content of the request_object to a file named "thinkpython.pdf"
    pdffile.write(request_object.content)

#Check that it downloaded the file to the current directory.
#%ls
````

### Example 6: Downloading and Saving PDF Files

PDF is a binary format, so we use `request.content` (not `.text`) to get the raw bytes. We can save these bytes directly to a file on disk.

````python
#To install pdftotext using pip, you need to first install poppler: brew install poppler
#%pip install pdftotext
import pdftotext

# Make sure you have downloaded the "thinkpython.pdf" file in your current folder
# http://www.greenteapress.com/thinkpython/thinkpython.pdf

# Load your PDF
with open('thinkpython.pdf', 'rb') as f:
    pdf = pdftotext.PDF(f)
    
# What kind of object is this?
#print(type(pdf)

# What are the methods and variables of this object.
#print(dir(pdf))

# Get more detail about how to use this object
# print(help(pdf))

# How many pages?
print("pages:", len(pdf))

# Iterate over all the pages
#for page in pdf:
#    print("\n=====newpage:=====\n", page)

# Read some individual pages
print("Page 0:\n", pdf[0])
print("Page 1:\n", pdf[1])

# Read all the text into one string
string_pdf = "\n\n".join(pdf)

# Print the first 500 characters in the string
print("\n\nThe first 500 symbols in the string:\n", string_pdf[:500])
````

### Using Command-Line Tools: wget

`wget` is a powerful command-line utility for downloading files and crawling websites. It's especially useful for:
- Downloading entire websites or sections of websites
- Following links recursively
- Mirroring websites for offline browsing

**Key parameters:**
- `-r`: Recursive download
- `-l [depth]`: Maximum recursion depth
- `--no-parent`: Don't ascend to parent directory
- `-E`: Adjust extension (e.g., .html)
- `-p`: Get all page requisites (images, CSS, etc.)

The *wget* tool is great for crawling entire or parts of websites. It recursively follows URLs up to given depth.
The example below downloads a part of the website locally, in a folder named *en.wikipedia.org*. The parameter -l tells wget to what depth it should follow URLs from the original URL. The parameter --no-parent tells wget to not download anything other than the given path. See http://linuxreviews.org/quicktips/wget/ for more details.

````python
#Crawl the website to depth 1. To stop downloading interrupt the kernel from the menu above.
# To install wget on Mac use: brew install wget
! wget -r -l 1 --no-parent https://en.wikipedia.org/wiki/Main_Page 
````

### Understanding JSON Structure

JSON (JavaScript Object Notation) is a lightweight data format that's easy for both humans and machines to read. It's the most common format for web APIs. JSON objects are similar to Python dictionaries, with nested key-value pairs.

Let's create a sample JSON file to understand its structure:

````python
json_string = """
{
    "glossary": {
        "title": "example glossary",
		"GlossDiv": {
            "title": "S",
			"GlossList": {
                "GlossEntry": {
                    "ID": "SGML",
					"SortAs": "SGML",
					"GlossTerm": "Standard Generalized Markup Language",
					"Acronym": "SGML",
					"Abbrev": "ISO 8879:1986",
					"GlossDef": {
                        "para": "A meta-markup language, used to create markup languages such as DocBook.",
						"GlossSeeAlso": ["GML", "XML"]
                    },
					"GlossSee": "markup"
                }
            }
        }
    }
}"""
with open("example.json", "w") as file:
    file.write(json_string)    
````

````python
# Run command "cat" to look at the file
# The sign ! tells Jupyter Notebook that the following is a Terminal command.
!cat example.json
````

### Reading and Parsing JSON Files

Once we have a JSON file, we can parse it using Python's `json` module. The parsed data becomes a nested Python dictionary that we can navigate using keys.

````python
json_data = json.load(open('example.json'))
#json_data looks like a nested Python dictionary
print(json_data)
````

````python
#We can refer to different fields of the json object
print(json_data['glossary']['title'])
print(json_data['glossary']['GlossDiv']['title'])
print(json_data['glossary']['GlossDiv']['GlossList']['GlossEntry']['ID'])
````

### Format: Parquet Files (.parquet)

Parquet is a columnar storage format optimized for big data analytics. It's becoming increasingly popular in data science and machine learning.

**Pros:**
- Excellent compression and query performance
- Preserves data types and schema
- Supports nested/complex data structures
- Very fast for analytical queries

**Cons:**
- Not human-readable
- Requires specialized libraries
- Slower write speed than CSV

````python
# Working with Parquet files
# Install if needed: pip install pyarrow or pip install fastparquet

import pandas as pd
import numpy as np

# Create sample data
df_parquet = pd.DataFrame({
    'id': range(1000),
    'timestamp': pd.date_range('2024-01-01', periods=1000, freq='h'),
    'value': np.random.randn(1000),
    'category': np.random.choice(['A', 'B', 'C'], 1000)
})

# Write to Parquet
df_parquet.to_parquet('data.parquet', compression='snappy')
print("Parquet file written")

# Read Parquet file
df_loaded = pd.read_parquet('data.parquet')
print("Shape:", df_loaded.shape)
print("Data types:\n", df_loaded.dtypes)
print(df_loaded.head())

# Read only specific columns (efficient for large files)
df_subset = pd.read_parquet('data.parquet', columns=['id', 'value'])
print("\nSubset shape:", df_subset.shape)

# Check file size comparison
import os
csv_size = os.path.getsize('data.csv') / (1024**2) if os.path.exists('data.csv') else 0
parquet_size = os.path.getsize('data.parquet') / (1024**2)
print(f"\nFile sizes - CSV: {csv_size:.2f}MB, Parquet: {parquet_size:.2f}MB")
````

### Format: JSONL (JSON Lines) Files

JSONL is a format where each line is a valid JSON object. It's ideal for large streaming datasets where you process one record at a time.

**Pros:**
- Handles very large datasets without loading entire file into memory
- Each line is a complete, independent JSON object
- Good for APIs that stream data
- Easy to process incrementally

**Cons:**
- Not as compact as Parquet
- Can't query without reading sequentially

````python
# Working with JSONL (JSON Lines) files

import json
import pandas as pd

# Create sample JSONL data
jsonl_records = [
    {'id': 1, 'name': 'Alice', 'score': 95.5, 'timestamp': '2024-01-01T10:00:00'},
    {'id': 2, 'name': 'Bob', 'score': 87.3, 'timestamp': '2024-01-01T11:00:00'},
    {'id': 3, 'name': 'Charlie', 'score': 92.1, 'timestamp': '2024-01-01T12:00:00'},
    {'id': 4, 'name': 'Diana', 'score': 88.9, 'timestamp': '2024-01-01T13:00:00'},
]

# Write to JSONL file (one JSON object per line)
with open('data.jsonl', 'w') as f:
    for record in jsonl_records:
        f.write(json.dumps(record) + '\n')
print("JSONL file written")

# Read JSONL file - Method 1: Load all at once
data = []
with open('data.jsonl', 'r') as f:
    for line in f:
        data.append(json.loads(line))

print("Records loaded:", len(data))
print("First record:", data[0])

# Read JSONL file - Method 2: Using pandas (efficient)
df_jsonl = pd.read_json('data.jsonl', lines=True)
print("\nDataFrame from JSONL:")
print(df_jsonl)

# Read JSONL - Method 3: Process incrementally (memory efficient for large files)
print("\nProcessing JSONL incrementally:")
with open('data.jsonl', 'r') as f:
    for i, line in enumerate(f):
        record = json.loads(line)
        print(f"  Record {i+1}: {record['name']} - Score: {record['score']}")
        # In practice, you'd process/aggregate records here for large files
````
